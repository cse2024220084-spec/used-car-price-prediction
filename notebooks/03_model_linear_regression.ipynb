## Setup

```python
import pandas as pd
import sys
from pathlib import Path

sys.path.append('../src')

from evaluate import evaluate_model, save_metrics
from utils import save_model
from sklearn.linear_model import LinearRegression
```

## Load Processed Data

```python
train_df = pd.read_csv('../data/processed/X_train.csv')
test_df = pd.read_csv('../data/processed/X_test.csv')

# Load target values
train_y = pd.read_csv('../data/processed/y_train.csv')['price']
test_y = pd.read_csv('../data/processed/y_test.csv')['price']

print('Train features shape:', train_df.shape)
print('Test features shape :', test_df.shape)
print('Train target shape  :', train_y.shape)
print('Test target shape   :', test_y.shape)
```

## TODO: One-hot encode categoricals, scale numeric features (see preprocessing.scale_features)

```python
X_train = train_df
X_test = test_df

y_train = train_y
y_test = test_y

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape", y_train.shape)
print("y_test shape", y_test.shape)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

## Evaluate

```python
metrics = evaluate_model(
    y_test,
    y_pred,
    "Linear Regression"
)

save_metrics(metrics)
metrics
```

## Save Model (joblib)Saves to `../models/linear_regression.joblib` so the FastAPI app can load it without retraining.

```python
save_model(model, 'linear_regression')
```

## Coefficient Interpretation- TODO: which features have the largest positive/negative coefficients?

```python
coefficients = pd.Series(
    model.coef_,
    index=X_train.columns
)
print("Largest positive coefficients:")
print(coefficients.sort_values(ascending=False).head())

print("\nLargest negative coefficients:")
print(coefficients.sort_values().head())
```
