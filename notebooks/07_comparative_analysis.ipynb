## Visual Comparison

```python
import matplotlib.pyplot as plt
import sys

try:
    metrics_df
except NameError:
    sys.path.append('../src')
    from evaluate import load_all_metrics
    metrics_df = load_all_metrics()

comparison_df = metrics_df.rename(columns={
    'model_name': 'model',
    'mse': 'MSE',
    'rmse': 'RMSE',
    'mae': 'MAE',
    'r2': 'R2'
}).copy()

comparison_df.set_index('model')[['RMSE', 'MAE']].plot(
    kind='bar',
    figsize=(8, 5),
    color=['#4C72B0', '#55A868']
)
plt.title('Model Comparison: RMSE & MAE')
plt.ylabel('Error')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
